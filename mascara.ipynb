```python
import io
from pathlib import Path
import pandas as pd


# Lê os dados (se fosse um arquivo real, use: df = pd.read_csv('seu_arquivo.csv'))
arquivo = Path("/home/wsl/gh/mestrado/claude9/DADOS/novoCj/APIs-CjAnterior-20260501-20260920-mes.csv")
if not arquivo.is_file():
    encontrados = list(Path.cwd().rglob("/home/wsl/gh/mestrado/claude9/DADOS/novoCj/APIs-CjAnterior-20260501-20260920-mes.csv"))
    if encontrados:
        arquivo = encontrados[0]
    else:
        raise FileNotFoundError(
            f"Não encontrei 'dados reduzido.csv' em {Path.cwd()}. "
            "Coloque o arquivo nessa pasta ou ajuste o caminho."
        )

df = pd.read_csv(arquivo)

# Mostra o DataFrame original
display(df)
```

```python
import hashlib

# Função que converte o valor para string e gera um hash SHA-256
def mascarar_hash(valor):
    valor_str = str(valor).encode('utf-8')
    semente = b"codigo-fixa42"
    return hashlib.sha256(semente + valor_str).hexdigest()

# Substitui CD_CLI pelo hash (não mantém o valor original)
df['CD_CLI'] = df['CD_CLI'].apply(mascarar_hash)

# Visualiza o DataFrame com a coluna mascarada
display(df)
```

```python
# Grava o resultado com CD_CLI mascarado ao lado do arquivo original
saida = arquivo.with_name(arquivo.stem + "_hash.csv")
df.to_csv(saida, index=False)
print(f"Arquivo gravado: {saida}")
```
